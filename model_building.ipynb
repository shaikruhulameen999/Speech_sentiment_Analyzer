```python
import torch

class SentimentDataset(torch.utils.data.Dataset):

    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {
            key: torch.tensor(val[idx])
            for key, val in self.encodings.items()
        }

        item["labels"] = torch.tensor(
            self.labels[idx],
            dtype=torch.long
        )

        return item

    def __len__(self):
        return len(self.labels)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=-1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="binary"
    )

    accuracy = accuracy_score(labels, predictions)

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

```python
import pandas as pd
from pathlib import Path

dataset_folder = Path(
    r"C:\Users\Dell\OneDrive\Desktop\Speech_to_Text_Sentiment\Dataset"
)

# Find files automatically
amazon_path = next(dataset_folder.rglob("amazon_cells_labelled.txt"))
imdb_path = next(dataset_folder.rglob("imdb_labelled.txt"))
yelp_path = next(dataset_folder.rglob("yelp_labelled.txt"))

# Load datasets
amazon = pd.read_csv(
    amazon_path,
    sep="\t",
    header=None,
    names=["text", "label"]
)

imdb = pd.read_csv(
    imdb_path,
    sep="\t",
    header=None,
    names=["text", "label"]
)

yelp = pd.read_csv(
    yelp_path,
    sep="\t",
    header=None,
    names=["text", "label"]
)

# Combine
df = pd.concat(
    [amazon, imdb, yelp],
    ignore_index=True
)

print("Amazon:", amazon.shape)
print("IMDB:", imdb.shape)
print("Yelp:", yelp.shape)
print("Final dataset:", df.shape)

print("\nFirst 5 rows:")
print(df.head())

print("\nLabel distribution:")
print(df["label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)

print("Training:", train_df.shape)
print("Testing:", test_df.shape)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

train_encodings = tokenizer(
    train_df["text"].tolist(),
    truncation=True,
    padding=True,
    max_length=128
)

test_encodings = tokenizer(
    test_df["text"].tolist(),
    truncation=True,
    padding=True,
    max_length=128
)

print("✅ Tokenization completed")
```

```python
import torch

class SentimentDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {
            key: torch.tensor(val[idx])
            for key, val in self.encodings.items()
        }
        item["labels"] = torch.tensor(
            self.labels[idx],
            dtype=torch.long
        )
        return item

    def __len__(self):
        return len(self.labels)


train_dataset = SentimentDataset(
    train_encodings,
    train_df["label"].tolist()
)

test_dataset = SentimentDataset(
    test_encodings,
    test_df["label"].tolist()
)

print("✅ PyTorch datasets created")
print("Train:", len(train_dataset))
print("Test:", len(test_dataset))
```

```python
from transformers import AutoModelForSequenceClassification

model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=2
)

print("✅ DistilBERT model loaded")
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./models/distilbert_sentiment",
    num_train_epochs=3,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    learning_rate=2e-5,
    weight_decay=0.01,
    logging_steps=50,
    eval_strategy="epoch",
    save_strategy="epoch",
    report_to="none"
)

print("✅ Training arguments ready")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset
)

print("✅ Trainer ready")
```

```python
trainer.train()
```

```python
results = trainer.evaluate()

print("✅ Evaluation completed")
print(results)
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, classification_report

predictions = trainer.predict(test_dataset)

pred_labels = np.argmax(
    predictions.predictions,
    axis=1
)

true_labels = test_df["label"].values

accuracy = accuracy_score(
    true_labels,
    pred_labels
)

print(f"Accuracy: {accuracy:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        true_labels,
        pred_labels,
        target_names=["Negative", "Positive"]
    )
)
```

```python
model_path = "./models/distilbert_sentiment"

trainer.save_model(model_path)
tokenizer.save_pretrained(model_path)

print("✅ Model saved successfully!")
print("Location:", model_path)
```
